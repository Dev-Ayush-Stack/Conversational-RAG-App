## Basic-Chat-Model Schema

```python
import os 
from dotenv import load_dotenv,find_dotenv
load_dotenv(find_dotenv())
```

```python
from langchain_openai import ChatOpenAI
llm=ChatOpenAI(
    base_url="https://api.groq.com/openai/v1",
    model="openai/gpt-oss-120b",
    api_key=os.getenv("Free_API_KEY"),
    temperature=0.0,
)
```

## Basic-Embedding-Model Schema


```python
from langchain_community.embeddings import HuggingFaceEmbeddings
embed_model=HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5",
)
```

```python
# Testing the chat model 
response=llm.invoke("What is the name of first women to get the nobel prize?")
print(response.content)
```

```python
# Testing the embedding model 
embeddings=embed_model.embed_query("What is the name of first women to get the nobel prize?")
print(embeddings[:5])
```
